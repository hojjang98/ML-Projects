# 평가 지표 정의하기

```python
def smape(gt, preds):
    gt= np.array(gt)
    preds = np.array(preds)
    v = 2 * abs(preds - gt) / (abs(preds) + abs(gt))
    score = np.mean(v) * 100
    return score
```

```python
def weighted_mse(alpha = 1):
    def weighted_mse_fixed(label, pred):
        residual = (label - pred).astype("float")
        grad = np.where(residual>0, -2*alpha*residual, -2*residual)
        hess = np.where(residual>0, 2*alpha, 2.0)
        return grad, hess
    return weighted_mse_fixed
```

```python
def custom_smape(preds, dtrain):
    labels = dtrain.get_label()
    return 'custom_smape', np.mean(2 * abs(preds - labels) / (abs(preds) + abs(labels))) * 100
```

```python
def calculate_day_values(dataframe, target_column, output_column, aggregation_func):
    result_dict = {}

    grouped_temp = dataframe.groupby(['building_number', 'month', 'day'])[target_column].agg(aggregation_func)

    for (building, month, day), value in grouped_temp.items():
        result_dict.setdefault(building, {}).setdefault(month, {})[day] = value

    dataframe[output_column] = [
        result_dict.get(row['building_number'], {}).get(row['month'], {}).get(row['day'], None)
        for _, row in dataframe.iterrows()
    ]

    
train['day_max_temperature'] = 0.0
train['day_mean_temperature'] = 0.0

calculate_day_values(train, 'temperature', 'day_max_temperature', 'max')
calculate_day_values(train, 'temperature', 'day_mean_temperature', 'mean')
calculate_day_values(train, 'temperature', 'day_min_temperature', 'min')

train['day_temperature_range'] = train['day_max_temperature'] - train['day_min_temperature']

calculate_day_values(test, 'temperature', 'day_max_temperature', 'max')
calculate_day_values(test, 'temperature', 'day_mean_temperature', 'mean')
calculate_day_values(test, 'temperature', 'day_min_temperature', 'min')

test['day_temperature_range'] = test['day_max_temperature'] - test['day_min_temperature']
```

### CDH(냉방도시) 변수

```python
def CDH(xs):
    cumsum = np.cumsum(xs - 26)
    return np.concatenate((cumsum[:11], cumsum[11:] - cumsum[:-11]))

def calculate_and_add_cdh(dataframe):
    cdhs = []
    for i in range(1, 101):
        temp = dataframe[dataframe['building_number'] == i]['temperature'].values
        cdh = CDH(temp)
        cdhs.append(cdh)
    return np.concatenate(cdhs)

train['CDH'] = calculate_and_add_cdh(train)
test['CDH'] = calculate_and_add_cdh(test)
```

## XGB 건물 유형별 단일모델

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import RepeatedKFold, KFold, RandomizedSearchCV
from sklearn.metrics import make_scorer
from sklearn.base import clone
from sklearn.compose import TransformedTargetRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor

RANDOM_SEED = 2023
np.random.seed(RANDOM_SEED)

def smape(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    denom = (np.abs(y_true) + np.abs(y_pred)) / 2
    diff  = np.abs(y_true - y_pred) / np.where(denom == 0, 1, denom)
    return float(np.mean(diff) * 100)

smape_scorer = make_scorer(smape, greater_is_better=False)
rkf = RepeatedKFold(n_splits=7, n_repeats=2, random_state=RANDOM_SEED)

try:
    type_list
except NameError:
    type_list = train['building_type'].unique().tolist()

xgb_space = {
    "regressor__n_estimators": np.arange(800, 2001, 200),
    "regressor__learning_rate": np.linspace(0.02, 0.08, 13),
    "regressor__max_depth": [3,4,5,6],
    "regressor__min_child_weight": [5,8,10,12],
    "regressor__subsample": np.linspace(0.65, 0.9, 6),
    "regressor__colsample_bytree": np.linspace(0.65, 0.9, 6),
    "regressor__reg_alpha": [0.5,1,2,4,8],
    "regressor__reg_lambda": [1,1.5,2,4,8],
    "regressor__gamma": [0,0.1,0.2,0.4],
}
lgbm_space = {
    "regressor__n_estimators": np.arange(800, 2001, 200),
    "regressor__learning_rate": np.linspace(0.02, 0.08, 13),
    "regressor__max_depth": [-1,4,5,6],
    "regressor__num_leaves": [31,63,127],
    "regressor__min_child_samples": [20,30,50,80],
    "regressor__subsample": np.linspace(0.65, 0.9, 6),
    "regressor__colsample_bytree": np.linspace(0.65, 0.9, 6),
    "regressor__reg_alpha": [0.5,1,2,4,8],
    "regressor__reg_lambda": [1,1.5,2,4,8],
}

answer_series = pd.Series(index=test.index, dtype=float)
oof_series    = pd.Series(index=X.index,    dtype=float)

SEEDS = [2021, 2022, 2023, 2024, 2025]

def _safe_sort_by_date(df):
    if 'date' in df.columns:
        out = df.sort_values('date')
    else:
        out = df.copy()
    return out

def _align_cols(train_df, test_df, drop_cols=('building_type',)):
    dd = train_df.copy()
    tt = test_df.copy()
    if 'building_number' in dd.columns:
        dd = pd.get_dummies(dd, columns=['building_number'], drop_first=False)
        tt = pd.get_dummies(tt, columns=['building_number'], drop_first=False)
        tt = tt.reindex(columns=dd.columns, fill_value=0)
    to_drop = [c for c in drop_cols if c in dd.columns]
    dd.drop(columns=to_drop, inplace=True, errors='ignore')
    tt.drop(columns=to_drop, inplace=True, errors='ignore')
    return dd, tt

def _trimmed_mean(weights, trim_ratio=0.1):
    w = np.array(weights, dtype=float)
    if len(w) < 5:
        return float(np.mean(w))
    k = int(np.floor(len(w) * trim_ratio))
    if k == 0:
        return float(np.mean(w))
    w_sorted = np.sort(w)
    w_trim = w_sorted[k:len(w)-k]
    return float(np.mean(w_trim))

for t in type_list:

    x = X[X.building_type == t].copy()
    y = Y[Y.building_type == t]["power_consumption"].copy()
    x_test = test_X[test_X.building_type == t].copy()

   
    x = _safe_sort_by_date(x)
    y = y.loc[x.index]

    x, x_test = _align_cols(x, x_test, drop_cols=('building_type',))

    xgb_base_ttr = TransformedTargetRegressor(
        regressor=XGBRegressor(tree_method="hist", random_state=RANDOM_SEED, n_jobs=-1),
        func=np.log1p, inverse_func=np.expm1
    )
    lgbm_base_ttr = TransformedTargetRegressor(
        regressor=LGBMRegressor(random_state=RANDOM_SEED, n_jobs=-1),
        func=np.log1p, inverse_func=np.expm1
    )

    rs_xgb = RandomizedSearchCV(
        estimator=xgb_base_ttr, param_distributions=xgb_space,
        n_iter=10, scoring=smape_scorer, cv=rkf, n_jobs=-1,
        verbose=0, refit=True, random_state=RANDOM_SEED
    )
    rs_xgb.fit(x, y)
    xgb_best_ttr = rs_xgb.best_estimator_
    print(f"[{t}] XGB BEST CV SMAPE: {-rs_xgb.best_score_:.4f}")

    rs_lgbm = RandomizedSearchCV(
        estimator=lgbm_base_ttr, param_distributions=lgbm_space,
        n_iter=10, scoring=smape_scorer, cv=rkf, n_jobs=-1,
        verbose=0, refit=True, random_state=RANDOM_SEED
    )
    rs_lgbm.fit(x, y)
    lgbm_best_ttr = rs_lgbm.best_estimator_
    print(f"[{t}] LGBM BEST CV SMAPE: {-rs_lgbm.best_score_:.4f}")

    kf = KFold(n_splits=7, shuffle=False)
    oof_xgb  = pd.Series(index=x.index, dtype=float)
    oof_lgbm = pd.Series(index=x.index, dtype=float)
    fold_infos = []

    xgb_reg_params = xgb_best_ttr.regressor.get_params()
    lgb_reg_params = lgbm_best_ttr.regressor.get_params()

    for tr_idx, va_idx in kf.split(x):
        X_tr, X_va = x.iloc[tr_idx], x.iloc[va_idx]
        y_tr, y_va = y.iloc[tr_idx], y.iloc[va_idx]

        
        fm_xgb = TransformedTargetRegressor(
            regressor=XGBRegressor(**xgb_reg_params),
            func=np.log1p, inverse_func=np.expm1
        )
        fm_lgb = TransformedTargetRegressor(
            regressor=LGBMRegressor(**lgb_reg_params),
            func=np.log1p, inverse_func=np.expm1
        )

        fm_xgb.fit(X_tr, y_tr)
        fm_lgb.fit(X_tr, y_tr)

        px = fm_xgb.predict(X_va)
        pl = fm_lgb.predict(X_va)

        px = np.clip(px, 0, None)
        pl = np.clip(pl, 0, None)

        oof_xgb.loc[X_va.index]  = px
        oof_lgbm.loc[X_va.index] = pl

        fold_infos.append({"idx": X_va.index, "y": y_va.values, "px": px, "pl": pl})

    
    w_list = []
    grid = np.linspace(0, 1, 101)
    for info in fold_infos:
        y_va, px, pl = info["y"], info["px"], info["pl"]
        best_w, best_sm = 0.5, 1e9
        for w in grid:
            s = smape(y_va, w*px + (1-w)*pl)
            if s < best_sm:
                best_sm, best_w = s, w
        w_list.append(best_w)

    
    w_mean_trim = _trimmed_mean(w_list, trim_ratio=0.10)
    w_std       = float(np.std(w_list))
    dist        = abs(w_mean_trim - 0.5)

    shrink = 0.12 + 0.40*dist + 0.75*w_std
    shrink = float(np.clip(shrink, 0.08, 0.35))

    final_w = (1 - shrink) * w_mean_trim + shrink * 0.5
    final_w = float(np.clip(final_w, 0.20, 0.80))

    print(f"[{t}] w_mean_trim={w_mean_trim:.4f}, w_std={w_std:.4f}, shrink={shrink:.3f} → final_w={final_w:.4f}")

    
    oof_blend = final_w * oof_xgb.values + (1 - final_w) * oof_lgbm.values
    oof_blend = np.clip(oof_blend, 0, None)
    oof_score = smape(y.values, oof_blend)
    print(f"[{t}] OOF Blend SMAPE (robust w): {oof_score:.4f}")

    test_preds_seeds = []
    for sd in SEEDS:
        xb = clone(xgb_best_ttr)
        lb = clone(lgbm_best_ttr)
        try:
            xb.regressor.set_params(random_state=sd)
        except Exception:
            pass
        try:
            lb.regressor.set_params(random_state=sd)
        except Exception:
            pass

        xb.fit(x, y)
        lb.fit(x, y)

        p_x = xb.predict(x_test)
        p_l = lb.predict(x_test)

        p = final_w * p_x + (1 - final_w) * p_l
        p = np.clip(p, 0, None)
        test_preds_seeds.append(p)

    test_pred = np.mean(test_preds_seeds, axis=0)
    answer_series.loc[x_test.index] = test_pred

    
    oof_series.loc[x.index] = oof_blend

total_smape = smape(Y['power_consumption'].values, oof_series.values.astype(float))
print(f"Total SMAPE : {total_smape:.6f}")

answer_df = pd.DataFrame({"answer": answer_series})
pred_df   = pd.DataFrame({"pred":   oof_series})
print(answer_df.head())
print(pred_df.head())
```
