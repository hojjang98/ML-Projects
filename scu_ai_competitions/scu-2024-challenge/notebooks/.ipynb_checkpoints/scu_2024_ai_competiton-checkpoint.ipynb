# Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from itertools import combinations
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from sklearn.metrics import f1_score
from sklearn.feature_selection import SelectPercentile
from lightgbm import LGBMClassifier, early_stopping
from tqdm.auto import tqdm

SEED = 42
DATA_PATH = "./"
```

# Load Data

```python
train = pd.read_csv(f"{DATA_PATH}train.csv")
test = pd.read_csv(f"{DATA_PATH}test.csv")
print(f"Train shape: {train.shape}, Test shape: {test.shape}")
```

# Missing Value Handling

```python
fill_values = {
    "나이": 0,
    "성별": "UNK",
    "TSH": train["TSH"].mean(),
    "FreeT3": train["FreeT3"].mean(),
    "FreeT4": train["FreeT4"].mean()
}

for col, val in fill_values.items():
    for df in [train, test]:
        df[col] = df[col].fillna(val)
```

# Binary Feature Conversion

```python
binary_cols = [c for c in train.columns if c.endswith("여부") or c.endswith("이력")]
for df in [train, test]:
    df[binary_cols] = (df[binary_cols] == "예").astype(int)
```

# Feature Engineering

```python
train_ft = train.iloc[:, 1:-1].copy()
test_ft = test.iloc[:, 1:].copy()

# simple aggregation
train_ft["yes_sum"] = train_ft[binary_cols].sum(axis=1)
test_ft["yes_sum"] = test_ft[binary_cols].sum(axis=1)

# interaction features
for n in [2, 3]:
    for comb in combinations(binary_cols, n):
        name = "_".join(comb)
        train_ft[name] = train_ft[list(comb)].sum(axis=1)
        test_ft[name] = test_ft[list(comb)].sum(axis=1)

# ratio feature
train_ft["FreeT4_FreeT3"] = train_ft["FreeT4"] / train_ft["FreeT3"]
test_ft["FreeT4_FreeT3"] = test_ft["FreeT4"] / test_ft["FreeT3"]
```

# One-Hot Encoding

```python
cat_cols = train_ft.select_dtypes("object").columns.tolist()
ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

train_ohe = pd.DataFrame(ohe.fit_transform(train_ft[cat_cols]), 
                         columns=ohe.get_feature_names_out(cat_cols))
test_ohe = pd.DataFrame(ohe.transform(test_ft[cat_cols]), 
                        columns=ohe.get_feature_names_out(cat_cols))

train_ft = pd.concat([train_ft.drop(columns=cat_cols), train_ohe], axis=1)
test_ft = pd.concat([test_ft.drop(columns=cat_cols), test_ohe], axis=1)
```

# Scaling

```python
scaler = MinMaxScaler()
train_ft = pd.DataFrame(scaler.fit_transform(train_ft), columns=train_ft.columns)
test_ft = pd.DataFrame(scaler.transform(test_ft), columns=test_ft.columns)

target = train["target"]
```

# Feature Selection (Percentile Search)

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
cv_scores = []

for p in tqdm(range(5, 96)):
    selector = SelectPercentile(percentile=p)
    X_selected = selector.fit_transform(train_ft, target)
    model = LGBMClassifier(random_state=SEED)
    scores = cross_val_score(model, X_selected, target, cv=cv, scoring="f1", n_jobs=-1)
    cv_scores.append([p, scores.mean()])

cv_scores = np.array(cv_scores)
best_percentile = cv_scores[np.argmax(cv_scores[:, 1]), 0]

plt.plot(cv_scores[:, 0], cv_scores[:, 1])
plt.xlabel("Percentile"); plt.ylabel("CV F1 Score")
plt.title("Feature Selection Performance"); plt.grid(True)
plt.show()

selector = SelectPercentile(percentile=best_percentile)
train_ft = pd.DataFrame(selector.fit_transform(train_ft, target))
test_ft = pd.DataFrame(selector.transform(test_ft))
```

# Model Training with K-Fold

```python
models, scores = [], []
for tr_idx, val_idx in cv.split(train_ft, target):
    X_tr, y_tr = train_ft.iloc[tr_idx], target.iloc[tr_idx]
    X_val, y_val = train_ft.iloc[val_idx], target.iloc[val_idx]

    model = LGBMClassifier(n_estimators=1000, random_state=SEED, verbosity=-1)
    model.fit(X_tr, y_tr, eval_set=[(X_val, y_val)], 
              callbacks=[early_stopping(300)])
    
    pred = model.predict(X_val)
    score = f1_score(y_val, pred)
    
    models.append(model)
    scores.append(score)

print("Validation F1 per Fold:", scores)
print("Average F1:", np.mean(scores))
```

# Inference & Submission

```python
pred_list = [m.predict_proba(test_ft)[:, 1] for m in models]
pred = np.mean(pred_list, axis=0)
submit = pd.read_csv(f"{DATA_PATH}sample_submission.csv")
submit["target"] = (pred > 0.5).astype(int)
submit.to_csv("AI_Competition_Submit.csv", index=False)
```
